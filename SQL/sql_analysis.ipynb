```python
# ==========================================
# Create SQLite Database
# ==========================================

import os
import sqlite3

# Create Database folder if it doesn't exist
os.makedirs("Database", exist_ok=True)

# Create SQLite Database
conn = sqlite3.connect("Database/google_play_store.db")

print("✅ Database Created Successfully!")
```

```python
# ==========================================
# Load Cleaned Datasets
# ==========================================

import pandas as pd

# Load cleaned datasets
apps_clean = pd.read_csv("Cleaned_Data/googleplaystore_cleaned.csv")
reviews_clean = pd.read_csv("Cleaned_Data/googleplaystore_user_reviews_cleaned.csv")

print("✅ Cleaned datasets loaded successfully!\n")

print("Apps Dataset Shape:", apps_clean.shape)
print("Reviews Dataset Shape:", reviews_clean.shape)

print("\nFirst 5 Rows of Apps Dataset")
print(apps_clean.head())
```

```python
# ==========================================
# Store Data into SQLite Database
# ==========================================

# Save Apps dataset
apps_clean.to_sql(
    "apps",
    conn,
    if_exists="replace",
    index=False
)

# Save Reviews dataset
reviews_clean.to_sql(
    "reviews",
    conn,
    if_exists="replace",
    index=False
)

print("="*60)
print("✅ Data Imported into SQLite Successfully!")
print("="*60)
```

```python
# ==========================================
# Verify Database Tables
# ==========================================

query = """
SELECT name
FROM sqlite_master
WHERE type='table';
"""

tables = pd.read_sql_query(query, conn)

print("="*60)
print("TABLES INSIDE DATABASE")
print("="*60)

print(tables)
```

```python
# ==========================================
# Count Records in Each Table
# ==========================================

apps_count = pd.read_sql_query(
    "SELECT COUNT(*) AS Total_Apps FROM apps;",
    conn
)

reviews_count = pd.read_sql_query(
    "SELECT COUNT(*) AS Total_Reviews FROM reviews;",
    conn
)

print("="*60)
print("APPS TABLE")
print("="*60)
print(apps_count)

print("\n")

print("="*60)
print("REVIEWS TABLE")
print("="*60)
print(reviews_count)
```

```python
# ==========================================
# SQL Query 31
# Rank Apps by Rating
# ==========================================

query = """
SELECT
    App,
    Category,
    Rating,
    RANK() OVER(ORDER BY Rating DESC) AS Rank_No
FROM apps
LIMIT 20;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ==========================================
# SQL Query 32
# Top 3 Apps per Category
# ==========================================

query = """
SELECT *
FROM (
    SELECT
        App,
        Category,
        Rating,
        ROW_NUMBER() OVER(
            PARTITION BY Category
            ORDER BY Rating DESC
        ) AS Row_Num
    FROM apps
)
WHERE Row_Num <= 3;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ==========================================
# SQL Query 33
# Apps Above Average Rating
# ==========================================

query = """
SELECT
    App,
    Rating
FROM apps
WHERE Rating >
(
    SELECT AVG(Rating)
    FROM apps
)
ORDER BY Rating DESC;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ==========================================
# SQL Query 34
# Highest Rated App per Category
# ==========================================

query = """
SELECT
    Category,
    MAX(Rating) AS Highest_Rating
FROM apps
GROUP BY Category
ORDER BY Highest_Rating DESC;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ==========================================
# SQL Query 35
# Join Apps and Reviews
# ==========================================

query = """
SELECT
    a.App,
    a.Category,
    a.Rating,
    r.Translated_Review,
    r.Sentiment
FROM apps a
INNER JOIN reviews r
ON a.App = r.App
LIMIT 20;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ==========================================
# SQL Query 36
# Count Reviews by Sentiment
# ==========================================

query = """
SELECT
    Sentiment,
    COUNT(*) AS Total
FROM reviews
GROUP BY Sentiment;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ==========================================
# SQL Query 37
# Average Sentiment Polarity
# ==========================================

query = """
SELECT
    ROUND(AVG(Sentiment_Polarity),3) AS Average_Polarity
FROM reviews;
"""

result = pd.read_sql_query(query, conn)

print(result)
```

```python
# ==========================================
# SQL Query 38
# Apps with Most Positive Reviews
# ==========================================

query = """
SELECT
    App,
    COUNT(*) AS Positive_Reviews
FROM reviews
WHERE Sentiment='Positive'
GROUP BY App
ORDER BY Positive_Reviews DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

print(result)
```
